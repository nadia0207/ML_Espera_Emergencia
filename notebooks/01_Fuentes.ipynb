### **1.1. Contexto del Proyecto**

Las urgencias hospitalarias son uno de los puntos de mayor presión del sistema sanitario. La experiencia del paciente en urgencias depende de múltiples factores: tiempo de espera, nivel de urgencia, disponibilidad de personal y recursos del centro.

**Objetivo del proyecto:** Predecir el nivel de satisfacción del paciente (escala 1-5)
a partir de variables operativas y contextuales de la visita a urgencias.

**Tipo de problema:** Clasificación multiclase supervisada (5 clases: 1, 2, 3, 4, 5)

---

### **1.2. Fuente**

- **Dataset:** ER Wait Time Dataset
- **Plataforma:** Kaggle
- **URL:** https://www.kaggle.com/datasets/rivalytics/er-wait-time
- **Formato:** CSV
- **Licencia:** Público para uso educativo y de investigación

El dataset recoge información de visitas a salas de urgencias hospitalarias,
incluyendo tiempos de espera por fase, características del hospital, perfil
de la visita y el resultado de satisfacción del paciente.

---

### **1.3. Descripcion General del Conjunto de Datos**

- **Filas**: 5.000 visitas de pacientes simuladas  
- **Columnas**: 20 características, incluyendo detalles del hospital, información de las visitas y resultados  
- **Periodo de tiempo**: Un año de datos simulados (2024)

---

### **1.4. Descripcion de Columnas**

| Nombre de la columna                  | Descripción                                                                 |
|--------------------------------------|-----------------------------------------------------------------------------|
| `Visit ID`                           | Identificador único de cada visita                                          |
| `Patient ID`                         | Identificador único de cada paciente                                        |
| `Hospital ID`                        | Identificador único de cada hospital                                        |
| `Hospital Name`                      | Nombre del hospital                                                         |
| `Region`                             | Clasificación como urbano o rural                                           |
| `Visit Date`                         | Fecha y hora de la visita del paciente                                      |
| `Day of Week`                        | Día de la semana (ej. lunes, martes)                                        |
| `Season`                             | Estación (invierno, primavera, verano, otoño)                               |
| `Time of Day`                        | Franja horaria (ej. madrugada, tarde)                                       |
| `Urgency Level`                      | Nivel de urgencia: crítico, alto, medio, bajo                               |
| `Nurse-to-Patient Ratio`             | Proporción de enfermeros por paciente                                       |
| `Specialist Availability`            | Número de especialistas disponibles                                         |
| `Facility Size (Beds)`               | Número de camas disponibles                                                 |
| `Time to Registration (min)`         | Tiempo para completar el registro                                           |
| `Time to Triage (min)`               | Tiempo para la evaluación de triaje                                         |
| `Time to Medical Professional (min)` | Tiempo hasta ser atendido por un profesional sanitario                      |
| `Total Wait Time (min)`              | Tiempo total desde llegada hasta atención                                   |
| `Patient Outcome`                    | Resultado: admitido, alta, se fue sin ser atendido                          |
| `Patient Satisfaction`               | Nivel de satisfacción (1 a 5)                                               |
